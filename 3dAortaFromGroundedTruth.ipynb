```python
import os
import nibabel as nib
import numpy as np
from plotly.graph_objs import *
from plotly.offline import iplot, plot
from plotly import figure_factory as FF
from skimage import measure
from PIL import Image
def make_mesh(image, threshold=-300, step_size=1, spacing=(1.0, 1.0, 1.0)):
    print("Transposing surface")
    img = nib.load(image)
    data = img.get_fdata()

    print("Calculating surface")
    # todo: соотношения настраиваются через spacing. Коэффициенты берутся Pixel Spacing Attribute и Slice Location Attribute
    verts, faces, norm, val = measure.marching_cubes(
        data,
        threshold,
        step_size=step_size,
        allow_degenerate=True,
        spacing=spacing,
    )
    # verts, faces, norm, val = measure.marching_cubes(p, threshold, step_size=step_size, allow_degenerate=True)
    return verts, faces


def plotly_3d(verts, faces, filename="untitled.html"):
    x, y, z = zip(*verts)

    print("Drawing")

    # Make the colormap single color since the axes are positional not intensity.
    #    colormap=['rgb(255,105,180)','rgb(255,255,51)','rgb(0,191,255)']
    colormap = ["rgb(236, 236, 212)", "rgb(236, 236, 212)"]

    fig = FF.create_trisurf(
        x=x,
        y=y,
        z=z,
        plot_edges=False,
        colormap=colormap,
        simplices=faces,
        backgroundcolor="rgb(64, 64, 64)",
        title="Interactive Visualization",
    )
    iplot(fig)
    plot(fig, filename=filename, auto_open=False)

v, f = make_mesh('D:/imagetbd/imageTBAD/imagetb/label_4.nii', 0)
```

```python
plotly_3d(v,f,'d:/imagetb.html')
```
